```python
def histogram(input_str):
    empty_dict = dict()
    
    for char in input_str:
        if char in empty_dict:
           empty_dict[char] +=1
        else:
            empty_dict[char] = 1

    return empty_dict

```

```python
histogram(['word', 'text', 'word', 'word', 'string'])
```

```python
def reverse_tuple(tuple_set1, tuple_set2):
  for pair in tuple_set1:
    reversed = pair[::-1]
    if reversed in tuple_set2:
      return True
  return False
```

```python
reverse_tuple({(1, 2), (3, 4)}, {(2, 1), (5, 6)})
```

```python
reverse_tuple({(1, 2), (3, 4)}, {(1, 2), (2, 3)}) 
```
