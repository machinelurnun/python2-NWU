```python
import time
```

```python
def pythagorean_triples_naive(n):
    """
    Prints the Pythagorean triples such that a + b + c == n
            :return: None.
    """
    # GENERATE all candidate triples (a, b, c) with a triple for loop
    for a in range(1, n):
        for b in range(1, n):
            for c in range(1, n):

                # TEST whether the candidate triple is indeed a solution
                if a + b + c == n and a**2 + b**2 == c**2:
                    print(a, b, c)
```

```python
def pythagorean_triples(n):
    """
    Prints the Pythagorean triples such that a + b + c == n
    :return: None.
    """

    # GENERATE all candidate a and b values, taking into consideration
    # that addition is commutative.
    for a in range(1, n):
        for b in range(a, n): # ensures a <= b

            # GENERATE the only possible value of c that corresponds to the
            # current choices of a and b.
            c = n - a - b

            # TEST whether the candidate triple is indeed a solution
            if a**2 + b**2 == c**2:
                print(a, b, c)
```

```python
print("Slow implementation:")

start = time.time()
pythagorean_triples_naive(1000)
end = time.time()
print(end - start)
```

```python
print("Faster implementation:")

start = time.time()
pythagorean_triples(10000)
end = time.time()
print(end - start)
```

```python
def palindromic_product(pINT):
    largest_product = 0
    upper_limit = pow(10,pINT)
    lower_limit = pow(10,pINT-1)

    for a in range(lower_limit, upper_limit):
        for b in range(a, upper_limit):
            product = a * b
            if product > largest_product:
                if str(product) == str(product)[::-1]:
                    largest_product = product
    return largest_product
                
```

```python
def primes_list(pINT):
    p_list = []
    for num in range(2, pINT+1):
            for i in range(2, num):
                if (num % i) == 0:
                    break
        else:
            p_list.append(num)
    print(p_list)
```
