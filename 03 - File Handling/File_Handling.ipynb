```python
def histogram(input_str):
    empty_dict = dict()
    
    for char in input_str:
        if char in empty_dict:
           empty_dict[char] +=1
        else:
            empty_dict[char] = 1

    return empty_dict

def anagrams(input_string):
    with open('words.txt', 'r') as RF:
        with open('output.txt', 'w') as OT:
            for line in RF:
                word = line.strip()
                if histogram(input_string) == histogram(word):
                    print(word)
                    OT.write(word + "\n")
            OT.close()
        RF.close()
anagrams('resent')

```

```python
def heteropalindromes():
    wordset = set()
    heteropalindrome = set()
    with open('words.txt', 'r') as RF:
        for line in RF:
            wordset.add(line.strip())
        for word in wordset:
            if word[::-1] in wordset:
                word_tuple = (word, word[::-1])
                heteropalindrome.add(word_tuple)
    print(heteropalindrome)
heteropalindromes()

```
