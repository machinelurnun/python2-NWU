## Introduction
The two data structures we've seen at this point, lists and strings, are both **sequences**.  The elements they contain are _ordered_ and use indices to locate and refer to them. 

In this lab, you'll learn about **dictionaries**, which are sometimes called **maps** or **associative arrays**, depending on the language you're working in.  They generalize the notion of lists by replacing the set of indices by a set of **keys** that can be just about _anything_.

Some books opens with an example of English-Spanish dictionary, but this overplays the dictionary analogy and actually hide some of the power/flexibility a Python dictionary affords.

## Creating Dictionaries


### Creating Non-Empty Dictionaries
> A **dictionary** is a data type defined by
> 1. an opening _curly_ bracket: `{`
> 2. an comma-separated list of **key-value pairs**, where a colon separates each key from it's corresponding value. E.g.
>     `key1:val1, key2:val2, ..., keyN:valN`
> 3. a closing  _curly_ bracket: `}`
>
> These are typically written across multiple lines as follows:
> ```python
> dictionary = {
>     key1:val1,
>     key2:val2,
>     ...,
>     keyN:valN
> }

## Adding/Replacing Key-Value Pairs (Updating)
Recall, to _update_ existing values in a list, as long as we knew it's _index_, we could ran something like:
> ```python
> lst[index] = newValue
> ```
Dictionaries use a similar syntax.  The only difference is that instead of specifying the _index_, we must specify the _key_ of the value we'd like to change:
> ```python
> dictionary[key] = newValue
> ```
What's more is that this technique doubles as the way we add new key-value pairs to the dictionary (so no `insert()` or `append()` methods are necessary).

### Another Import Example: Duplicate Keys
What happens when you run the cell below:

````python
print(mckainsFavoriteThings)
mckainsFavoriteThings["food"] = "yellow curry"
print(mckainsFavoriteThings)
````

If you specify an existing key, it will update the corresponding value, instead of adding a new key value pair. **The set of keys must be distinct**.

This is because a dictionary is like a **function**; the keys are the inputs, the values are the outputs.  In the mathematical sense of the word, a _function_ takes each input to a _unique_ output; the input `"food"` cannot have 2 different output values (`"thai"` _and_ `"yellow curry"`), so the latest one is used. Even lists, the closest relative of a dictionary, only allow a single value per index.

## _Safely_ Viewing Values: The `get()` Method
We saw early that if the key doesn't exist in the list of keys, then trying to access the corresponding value yields an error. This means we have to make sure the key exists before each retrieve the value for that key, otherwise we risk catching that error.

An alternative is the `.get()` method.  This method takes 2 parameters; the key we want to look up, and _the default value that should return if the key is not found._  
For instance:

````python
d = {"age" : 1}
d["numberOfChildren"] # the same error as before"
````

That prints the same error we say before, since "y" is not in the list of keys.  However:

````python
d.get("numberOfChildren", -1)
d.get("age")
````

The command above says "look for a key that is `numberOfChildren`; if it exists returns it's value, otherwise return -1".  It's always a good idea to choose default values that convey some sort of meaning.  Having 5 as the "default number of children" is probably a bad idea, because that could be the actual number of children, and you'd never know.  Using `-1` however implies the information wasn't available for updating.

## Activity #1
Write a function `letterFrequencies` which takes a single parameter, a string, determines the number of times each letter appears in the string.  For instance `letterFrequencies('aabaacb')` should **return** an "object" that conveys the following information: 
```
The letter "a" appears 4 times.
The letter "b" appears 2 times.
The letter "c" appears 1 time.
```

You could do this using variables or a list, but how many branches will your conditional need to make sure the proper variable/list element gets updated?

Instead of either of those approaches (called **implementations**) we'll use a dictionary.
* Your function should first create an empty dictionary.
* It should then loop through all of the characters in the input `string`.
* If a character is **not** in the list of keys yet, then it should be added (what `value` should it get?)
* Otherwise, the value corresponding to the letter should be incremented by 1.
* Don't forget to return (not print) the final dictionary at the end.

````python
# your code here
def letterFrequencies(pString):
    dicta = {}
    for character in pString:
        if character not in dicta.keys():
            dicta[character] = 1
        elif character in dicta.keys():
            dicta[character] += 1
    return dicta

dictionary = letterFrequencies('aabaacb')
for character in dictionary.keys():
    print(f"The letter {character} appears {dictionary[character]} times.")

        
````
